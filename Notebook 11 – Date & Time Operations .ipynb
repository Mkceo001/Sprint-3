# Date & Time Operations

## Datetime
What is Datetime?

Datetime represents a specific date and time.
#### Example:
- Date: 2026-10-14
- Time: 10:30:00
- Datetime: 2026-10-14 10:30:00
  
Pandas uses the datetime64 data type to store date-and-time values efficiently.
### Exampl:
 Creating datetime values

```python
import pandas as pd

date = pd.to_datetime("2026-10-14")

print(date)
print(type(date))
```

## Timestamp
What is a Timestamp?

A Timestamp represents one specific point in time. In Pandas, it is the object commonly used for individual datetime values.
#### Example
A timestamp can represent when:
- A customer placed an order.
- An employee logged in.
- A payment was completed.
- A sensor recorded a measurement.
### Example: 
Creating a Timestamp


```python
import pandas as pd

ts = pd.Timestamp("2026-10-14 09:30:45")

print(ts)
print("Year:", ts.year)
print("Month:", ts.month)
print("Day:", ts.day)
print("Hour:", ts.hour)
print("Minute:", ts.minute)
print("Second:", ts.second)
```

## Date Parsing
What is Date Parsing?

Date parsing means converting date information from text into a proper datetime format.
###### Datasets often store dates in different formats:
- 14-10-2026
- 10/14/2026
- 2026-10-14
- 14 Oct 2026

These may be stored as strings. Converting them to datetime makes sorting, filtering, and calculation easier.
### Example :
Convert strings into dates

```python
import pandas as pd

df = pd.DataFrame({
    "Order_ID": [101, 102, 103],
    "Order_Date": [
        "2026-10-10",
        "2026-10-12",
        "2026-10-14"
    ]
})

df["Order_Date"] = pd.to_datetime(df["Order_Date"])

print(df)
print(df.dtypes)
```

 ## Timedelta
What is Timedelta?

A Timedelta represents a duration or difference between two points in time.
#### Example:
- 5 days between an order and delivery.
- 2 hours between login and logout.
- 30 minutes spent processing a transaction.
### Example:
Adding and subtracting time

```python
import pandas as pd

start_date = pd.Timestamp("2026-10-14")
duration = pd.Timedelta(days=5)

end_date = start_date + duration

print("Start:", start_date)
print("Duration:", duration)
print("End:", end_date)
```

## Time Zones
What is a Time Zone?

A time zone represents the local time rules for a region.
#### Example
The same instant can be represented as:
- India: IST (UTC+05:30)
- London: UTC or British local time, depending on the date
- New York: UTC−05:00 or UTC−04:00, depending on daylight saving time
  
Time-zone handling is important when an application serves users in different countries.
### Example:
Create a timezone-aware timestamp

```python
import pandas as pd

india_time = pd.Timestamp(
    "2026-10-14 10:00:00",
    tz="Asia/Kolkata"
)

print(india_time)
```

## Extracting Date Features
What does feature extraction mean?

Feature extraction means creating useful columns from existing data.
#### Example
From 2026-10-14 17:30:00, we can extract:
- Year: 2026
- Month: 10
- Day: 14
- Hour: 17
- Day of week: 2 (Wednesday, where Monday is 0)
- Quarter: 4
- 
This is especially useful for data analysis and machine learning.
### Example:
Extract date features

```python
import pandas as pd

df = pd.DataFrame({
    "Order_Date": [
        "2026-10-12 09:30:00",
        "2026-10-13 14:45:00",
        "2026-10-14 17:30:00"
    ]
})

df["Order_Date"] = pd.to_datetime(df["Order_Date"])

df["Year"] = df["Order_Date"].dt.year
df["Month"] = df["Order_Date"].dt.month
df["Day"] = df["Order_Date"].dt.day
df["Hour"] = df["Order_Date"].dt.hour
df["Day_of_Week"] = df["Order_Date"].dt.dayofweek
df["Day_Name"] = df["Order_Date"].dt.day_name()
df["Quarter"] = df["Order_Date"].dt.quarter
df["Is_Weekend"] = df["Order_Date"].dt.dayofweek >= 5

print(df)
```
