# Performance Optimization

## Vectorization
What is vectorization?

Vectorization means performing an operation on an entire NumPy array or Pandas column at once instead of processing each value individually using a Python loop.
### Example:

```python

import pandas as pd

df = pd.DataFrame({
    "Name": ["Arun", "Bala", "Kumar", "Ravi"],
    "Salary": [20000, 25000, 30000, 35000]
})

df["New_Salary"] = df["Salary"] * 1.10

# Optional: round to 2 decimal places
df["New_Salary"] = df["New_Salary"].round(2)

print(df)
```

## apply() vs Loops
Pandas provides apply() to apply a function to rows or columns. It can make code cleaner than writing an explicit loop, but it is not automatically faster than a loop in every situation.
### Example : Loop

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Arun", "Bala", "Kumar", "Ravi"],
    "Marks": [85, 45, 92, 30]
})

results = []

for mark in df["Marks"]:
    if mark >= 50:
        results.append("Pass")
    else:
        results.append("Fail")

df["Result"] = results

print(df)
```

### Example : Using apply()

```python
def check_result(mark):
    if mark >= 50:
        return "Pass"
    else:
        return "Fail"

df["Result"] = df["Marks"].apply(check_result)

print(df)
```

## Memory Optimization
What is memory optimization?

Memory optimization means reducing the amount of RAM used by a DataFrame while keeping the information and results correct.

Pandas data types include:
- int64: 64-bit integers
- int32: 32-bit integers
- float64: 64-bit decimal numbers
- float32: 32-bit decimal numbers
- object: often used for text
- category: efficient for repeated labels
  ### Example:

```python
import pandas as pd

df = pd.DataFrame({
    "Employee_ID": [101, 102, 103, 104, 105],
    "Age": [22, 25, 28, 30, 35],
    "Salary": [20000.5, 25000.5, 30000.5, 35000.5, 40000.5],
    "Department": [
        "IT", "HR", "IT", "Sales", "HR"
    ]
})

print(df.dtypes)

print(df.memory_usage(deep=True))
```

## Category Data Type
What is category?


The Pandas category data type is useful when a column contains a limited number of distinct values repeated many times.
### Example
A dataset may contain 100,000 employee records but only four departments:
- IT
- HR
- Sales
- Finance

```python
import pandas as pd

df = pd.DataFrame({
    "Department": [
        "IT", "HR", "IT", "Sales", "HR",
        "Finance", "IT", "Sales", "HR", "IT"
    ]
})

print(df.dtypes)

print(
    "Memory:",
    df["Department"].memory_usage(deep=True),
    "bytes"
)
```

```python
#Convert to category 
df["Department"] = df["Department"].astype("category")

print(df.dtypes)

print(
    "Memory:",
    df["Department"].memory_usage(deep=True),
    "bytes"
)
```
