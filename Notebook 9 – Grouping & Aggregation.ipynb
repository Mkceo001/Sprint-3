# Grouping & Aggregation

## Sample dataset

```python
import pandas as pd

data = {
    "Name": ["Arun", "Bala", "Kumar", "Priya", "Divya",
             "Ravi", "Meena", "Suresh"],
    "Department": ["IT", "HR", "IT", "HR", "IT",
                   "Finance", "Finance", "IT"],
    "Gender": ["Male", "Male", "Male", "Female", "Female",
               "Male", "Female", "Male"],
    "Salary": [30000, 25000, 40000, 28000, 35000,
               32000, 38000, 45000],
    "Experience": [1, 2, 4, 2, 3, 3, 5, 6]
}

df = pd.DataFrame(data)

print(df)
```

## groupby() — Grouping data
What is groupby()?

The Pandas groupby() method groups rows that have the same value in a particular column.
For example, we can group employees by department to analyze the IT, HR, and Finance teams separately.
###### Example:
A company wants to know how many employees work in each department.
### Example: 
Group employees by department

```python
grouped = df.groupby("Department")

print(grouped.size())
```

## agg() — Aggregation
What is agg()?

Aggregation means calculating summary values from a group of rows.
The agg() method allows us to calculate multiple statistics, such as sum, mean, minimum, maximum, and count, in one operation.
### Example: 
Calculate multiple statistics

```python
result = df.groupby("Department")["Salary"].agg(
    ["sum", "mean", "min", "max", "count"]
)

print(result)
```

## transform() — Calculate group values for every row
What is transform()?

The transform() method performs a calculation for each group but returns a result with the same number of rows as the original DataFrame.
This is the main difference between transform() and a normal aggregation.
### Example: 
Add the department's average salary

```python
df["Department_Avg_Salary"] = (
    df.groupby("Department")["Salary"].transform("mean")
)

print(df[["Name", "Department", "Salary",
          "Department_Avg_Salary"]])
```

## filter() — Keep or remove entire groups
What is filter()?

The filter() method keeps groups that satisfy a condition and removes the other groups.
Unlike ordinary row filtering, it makes a decision about the whole group.
### Example:
Keep departments with more than two employees

```python
result = df.groupby("Department").filter(
    lambda group: len(group) > 2
)

print(result[["Name", "Department", "Salary"]])
```

## Pivot Tables — Summarize data across rows and columns
What is a pivot table?

A pivot table summarizes data using one category for rows, another category for columns, and a calculation for the values.
It is similar to the PivotTable feature in Microsoft Excel.
#### Real-life example:
A manager wants to compare the average salary of male and female employees across departments.
#### Example:
Average salary by department and gender

```python
pivot = pd.pivot_table(
    df,
    values="Salary",
    index="Department",
    columns="Gender",
    aggfunc="mean"
)

print(pivot)
```

## crosstab() — Count relationships between categories
What is crosstab()?

The pd.crosstab() function creates a frequency table that counts how often combinations of categorical values occur.
It is especially useful for understanding the relationship between two categorical variables.
### Example: 
Count employees by department and gender

```python
result = pd.crosstab(
    df["Department"],
    df["Gender"]
)

print(result)
```
