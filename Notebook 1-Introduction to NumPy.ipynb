# Introduction to NumPy

## What is NumPy?
NumPy stands for Numerical Python. It is a Python library used to work with numerical data, perform mathematical calculations, and manage arrays efficiently.
It provides a special data structure called an ndarray, which can store numbers in one or more dimensions.
### Example
Imagine a teacher has marks for five students: 70, 80, 65, 90, and 85. NumPy can store these marks in an array and calculate their average quickly.

```python
import numpy as np

marks = np.array([70, 80, 65, 90, 85])

print("Marks:", marks)
print("Average:", np.mean(marks))
```

## Why NumPy?
NumPy makes numerical calculations easier and more efficient. It supports operations on entire arrays without requiring a separate Python loop for every element. It also provides mathematical, statistical, and linear algebra functions.

## NumPy Arrays (ndarray)
An ndarray is NumPy's main array object. It stores elements in an organized structure and supports one-dimensional, two-dimensional, and higher-dimensional data.
### Example
A list of student marks represents one dimension. A table of marks for several students across different subjects represents two dimensions.

```python
import numpy as np

one_d = np.array([10, 20, 30])
two_d = np.array([[10, 20], [30, 40]])

print("One-dimensional array:")
print(one_d)

print("Two-dimensional array:")
print(two_d)

print("Type:", type(one_d))
```

## Creating Arrays
NumPy provides different functions to create arrays depending on the data and calculation required. We can create arrays from lists or generate arrays filled with zeros, ones, or sequences of numbers.
## Example
A machine learning model might need an array of zeros to initialize values or a sequence of numbers to represent time steps.

```python
import numpy as np

a = np.array([1, 2, 3])
b = np.zeros(4)
c = np.ones(3)
d = np.arange(1, 6)
e = np.linspace(0, 1, 5)

print("Array:", a)
print("Zeros:", b)
print("Ones:", c)
print("Range:", d)
print("Evenly spaced:", e)
```

## Dimensions
The dimension of an array describes the number of axes it has. A one-dimensional array has one axis, a two-dimensional array has two axes, and a three-dimensional array has three axes.
### Example
- 1D: Student marks.
- 2D: A table containing students and their marks.
- 3D: A collection of multiple two-dimensional tables.


```python
import numpy as np

a = np.array([10, 20, 30])
b = np.array([[1, 2], [3, 4]])
c = np.array([[[1, 2], [3, 4]]])

print("1D dimensions:", a.ndim)
print("2D dimensions:", b.ndim)
print("3D dimensions:", c.ndim)
```

 ## Shape
The shape of an array tells us how many elements it contains along each dimension. It is represented as a tuple.
## Example
A table with three rows and two columns has the shape (3, 2).

```python
import numpy as np

data = np.array([[10, 20], [30, 40], [50, 60]])

print("Shape:", data.shape)
```

## Size

The size of an array is the total number of elements it contains, regardless of how many dimensions it has.
### Example
A table containing three rows and four columns has 12 elements.

```python
import numpy as np

data = np.array([[1, 2, 3], [4, 5, 6]])

print("Total elements:", data.size)
```

## Data Types
The data type, or dtype, tells us what kind of values an array stores, such as integers, decimal numbers, or Boolean values. NumPy arrays usually use a common data type for all their elements.
## Example
Student counts can be stored as integers, while average marks may require decimal values.

```python
import numpy as np

a = np.array([10, 20, 30])
b = np.array([1.5, 2.5, 3.5])
c = np.array([True, False, True])

print("Integer type:", a.dtype)
print("Float type:", b.dtype)
print("Boolean type:", c.dtype)
```

## Indexing
Indexing is used to access a specific element in an array. NumPy uses zero-based indexing, which means the first element has index 0.
### Example
If a product price array contains [100, 200, 300], index 1 refers to the second price, which is 200.

```python
import numpy as np

prices = np.array([100, 200, 300, 400])

print("First price:", prices[0])
print("Second price:", prices[1])
print("Last price:", prices[-1])
```

## Slicing
Slicing extracts a portion of an array. The syntax array[start:stop] includes the starting index but excludes the stopping index.
### Example
A teacher wants to retrieve marks for the second, third, and fourth students from a list of marks.

```python
import numpy as np

marks = np.array([70, 80, 65, 90, 85])

print(marks[1:4])
print(marks[:3])
print(marks[2:])
```

## Reshaping
Reshaping changes the dimensions of an array without changing its total number of elements. The new shape must be compatible with the existing number of elements.
### Example
Six numbers can be arranged as a one-dimensional array or as a table containing two rows and three columns.eshaping


```python
import numpy as np

data = np.array([1, 2, 3, 4, 5, 6])

reshaped = data.reshape(2, 3)

print("Original:", data)
print("Reshaped:")
print(reshaped)
```

## Flatten
Flattening converts a multidimensional array into a one-dimensional array. The elements remain in their existing order.
### Example
A two-dimensional table can be converted into a single sequence of values.

```python
import numpy as np

data = np.array([[1, 2, 3], [4, 5, 6]])

flat = data.flatten()

print("Original:")
print(data)
print("Flattened:", flat)
```

## Copy vs View
A copy creates a separate array with its own data. A view provides another way to access the same underlying data. Changes to shared data can affect both the original array and its view.
### Example
A copy is like making a photocopy of a document. Editing the photocopy does not change the original. A view is like looking at the same document through another window.

```python
import numpy as np

original = np.array([10, 20, 30])

copied = original.copy()
viewed = original.view()

copied[0] = 100
viewed[1] = 200

print("Original:", original)
print("Copy:", copied)
print("View:", viewed)
```

## Broadcasting
Broadcasting allows NumPy to perform operations on arrays with compatible shapes, even when their dimensions are different. NumPy automatically expands the smaller shape conceptually without necessarily creating a full copy of the data.
###  Example
A shop increases every product price by the same amount. Instead of storing the increase separately for each product, we can add one number to the entire array.

```python
import numpy as np

prices = np.array([100, 200, 300])

updated_prices = prices + 50

print("Original prices:", prices)
print("Updated prices:", updated_prices)
```

## Array Operations
Array operations allow us to perform arithmetic and comparison operations on array elements. NumPy applies many of these operations element by element, making numerical calculations simple.
### Example
A business wants to calculate the total price after applying a discount to several products.

```python
import numpy as np

prices = np.array([100, 200, 300])
discounts = np.array([10, 20, 30])

final_prices = prices - discounts

print("Final prices:", final_prices)
print("Total:", np.sum(final_prices))
print("Average:", np.mean(final_prices))
print("Above 200:", final_prices > 200)
```

## Universal Functions (ufuncs)
Universal functions, or ufuncs, are NumPy functions designed to perform operations element by element on arrays. Examples include square root, exponential, sine, and absolute value.
### Example
A data analyst may need to calculate the square root of several numerical measurements. A ufunc performs this calculation on every element.

```python
import numpy as np

values = np.array([4, 9, 16, 25])

print("Square root:", np.sqrt(values))
print("Square:", np.square(values))
print("Absolute values:", np.abs([-5, -2, 3]))
print("Exponential:", np.exp([0, 1]))
```
