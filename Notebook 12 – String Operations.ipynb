# String Operations

### sample DataFrame

```python
import pandas as pd

df = pd.DataFrame({
    "Customer": ["  MUGESH  ", "ramya", "  ARUN"],
    "City": ["CHENNAI", "madurai", "TIRUNELVELI"],
    "Email": [
        "Mugesh@Gmail.com",
        "ramya@yahoo.com",
        "arun@gmail.com"
    ]
})

print(df)
```

### lower() 
Convert text to lowercase.
The lower() method converts all uppercase letters into lowercase letters.
### Example:

```python
df["Customer_Lower"] = df["Customer"].str.lower()

print(df[["Customer", "Customer_Lower"]])
```

## upper() 
Convert text to uppercase.
The upper() method converts lowercase letters into uppercase letters.
### Example:

```python
df["City_Upper"] = df["City"].str.upper()

print(df[["City", "City_Upper"]])
```

## strip() 
Remove extra spaces.
The strip() method removes whitespace from the beginning and end of a string. It does not remove spaces between words.
### Example

```python
names = pd.Series([
    "  Mugesh  ",
    " Ramya",
    "Arun   "
])

clean_names = names.str.strip()

print(clean_names)
```

## split() 
Divide a string into parts.
The split() method divides a string into smaller pieces using a separator.
### Example

```python
names = pd.Series([
    "Mugesh Kannan",
    "Ramya Devi",
    "Arun Kumar"
])

print(names.str.split(" "))
```

```python
emails = pd.Series([
    "mugesh@gmail.com",
    "ramya@yahoo.com",
    "arun@outlook.com"
])

print(emails.str.split("@"))
```

## contains() 
Search for text.
The contains() method checks whether a string contains a particular word, character, or pattern.
It returns Boolean values: True or False.
### Example:

```python
comments = pd.Series([
    "Python is easy",
    "I like Java",
    "Python is powerful",
    "I am learning SQL"
])

result = comments.str.contains("Python")

print(result)
```

## replace() 
Replace text.
The replace() method substitutes specified text with new text.
### Example

```python
comments = pd.Series([
    "Python is easy",
    "Python is popular",
    "Python is powerful"
])

result = comments.str.replace("Python", "Pandas")

print(result)
```

## Regular Expressions (Regex)
A Regular Expression (Regex) is a pattern used to search, identify, extract, or replace text based on specific rules.
In Pandas, regex is useful when normal string methods are not enough to identify a pattern.
#### Example, 
you might want to:
- Extract email addresses from text.
- Find phone numbers.
- Remove special characters.
- Identify strings containing only digits.
- Replace multiple spaces with a single space.
  
Python provides the re module, and Pandas string methods such as .str.contains(), .str.replace(), and .str.extract() support regular expressions.
