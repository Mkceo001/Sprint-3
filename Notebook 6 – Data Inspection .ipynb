# Data Inspection 


```python
import pandas as pd

# Create the sample dataset
data = {
    "Name": ["Arun", "Priya", "Kumar", "Divya",
             "Arun", "Vijay", "Meena", "Priya"],
    "Age": [25, 28, 32, 24, 25, 30, 27, 28],
    "Department": ["IT", "HR", "IT", "Finance",
                   "IT", "HR", "Finance", "HR"],
    "Salary": [30000, 35000, 45000, 28000,
               30000, 40000, 32000, 35000],
    "Experience": [2, 4, 8, 1, 2, 6, 3, 4]
}

# Create the DataFrame
df = pd.DataFrame(data)

# Display the dataset
print(df)
```

## head()

Definition: The head() method displays the first few rows of a DataFrame.

It is useful when we want to quickly understand what a dataset looks like after loading it from a CSV file or database.
### Example

```python
print(df.head())
```

## tail() — View the Last Rows
Definition: The tail() method displays the last few rows of a DataFrame.

It is useful for checking the end of a dataset and verifying whether the last records were loaded correctly.
### Example

```python
print(df.tail())
```

```python
print(df.tail(2))
```

## sample() — View Random Rows
Definition: The sample() method selects random rows from a DataFrame.

It helps us inspect different parts of a dataset instead of always looking at the beginning or end.
### Example

```python
print(df.sample(3))
```

```python
print(df.sample(3, random_state=42))
```

## info() — Get Dataset Information
Definition: The info() method displays a summary of the DataFrame, including column names, non-null counts, data types, and memory usage.
### Example

```python
df.info()
```

## describe() — Statistical Summary
Definition: The describe() method generates descriptive statistics for numerical columns by default.

It helps us understand the distribution and range of numerical data.
### Exanmple

```python
print(df.describe())
```

## shape — Find Rows and Columns
Definition: shape is a DataFrame attribute that returns the number of rows and columns as a tuple.
### Example

```python
print(df.shape)
```

## columns — Display Column Names
Definition: The columns attribute returns the names of all columns in a DataFrame.
### Example:

```python
print(df.columns)
```

## dtypes — Check Data Types
Definition: The dtypes attribute displays the data type of each DataFrame column.
### Example:

```python
print(df.dtypes)
```

## memory_usage() — Check Memory Consumption
Definition: The memory_usage() method returns the memory used by each DataFrame column, measured in bytes.
### Example:

```python
print(df.memory_usage())
```

## nunique() — Count Unique Values
Definition: The nunique() method counts the number of distinct values in a column or across each column.
By default, missing values are excluded.
### Example:

```python
print(df.nunique())
```

## value_counts() — Count How Often Values Appear
Definition: The value_counts() method counts the frequency of each distinct value in a Series.
It is particularly useful for categorical columns.
### Example:

```python
print(df["Department"].value_counts())
```
