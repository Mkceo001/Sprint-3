# Notebook 7 – Selecting & Filtering Data

## sample dataset
We will use an employee dataset throughout this notebook.

```python
import pandas as pd

data = {
    "Employee_ID": [101, 102, 103, 104, 105, 106],
    "Name": ["Arun", "Bala", "Charan", "Divya", "Ezhil", "Farah"],
    "Age": [23, 28, 35, 26, 42, 31],
    "Department": [
        "IT", "HR", "IT", "Finance", "HR", "IT"
    ],
    "Salary": [25000, 35000, 60000, 30000, 50000, 45000],
    "Experience": [1, 3, 8, 2, 10, 5]
}

df = pd.DataFrame(data)

print(df)
```

## loc[] – Label-based selection
What is loc[]?

loc[] is used to select rows and columns using their index labels and column names.

Think of it like finding an employee using the employee's row label and choosing the information you want to see.
### Example

```python
print(df.loc[0])
```

## iloc[] – Integer-position-based selection
What is iloc[]?

iloc[] selects rows and columns based on their numerical positions, starting from 0.
Imagine a table where you choose the first row, second row, or third column by position rather than by its name.
### Example

```python
print(df.iloc[0])
```

## Boolean Indexing
What is Boolean Indexing?

Boolean indexing means filtering rows using a condition that produces True or False.
- True: Keep the row.
- False: Exclude the row.
For example, suppose we want employees whose salary is greater than ₹40,000. Python checks every employee's salary and selects only matching records.
### Example
 Salary greater than ₹40,000

```python
print(df["Salary"] > 40000)
```

## Filtering Rows
What is row filtering?

Filtering rows means displaying only the records that satisfy specific conditions.
Boolean indexing is one way to filter rows. Here, we will practise common business examples.
### Example: 
Find employees in the HR department

```python
hr_employees = df[df["Department"] == "HR"]

print(hr_employees)
```

## query() – Filter using a readable expression
What is query()?

The query() method filters a DataFrame using a string expression. It is often easier to read when the condition involves several columns.
### Example: 
Salary greater than ₹40,000

```python
result = df.query("Salary > 40000")

print(result)
```

## isin() – Match values from a list
What is isin()?

The isin() method checks whether each value belongs to a specified list of values.
It is useful when you want to select records matching one of several possible values.
### Example: 
Select IT and HR employees

```python
result = df[df["Department"].isin(["IT", "HR"])]

print(result[["Name", "Department"]])
```

### between() – Filter values within a range
What is between()?

The between() method checks whether a value falls within a specified range.
By default, both the lower and upper boundaries are included.
### Example: 
Employees aged 25 to 35

```python
result = df[df["Age"].between(25, 35)]

print(result[["Name", "Age"]])
```

## Sorting Data
What is sorting?

Sorting means arranging records in a particular order, such as lowest to highest salary or highest to lowest salary.
Pandas provides sort_values() to sort by column values and sort_index() to sort by index labels.
### Example: 
Sort by salary in ascending order

```python
result = df.sort_values(by="Salary")

print(result[["Name", "Salary"]])
```
