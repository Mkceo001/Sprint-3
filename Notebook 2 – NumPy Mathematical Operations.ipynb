# NumPy Mathematical Operations

## Sum
The sum operation calculates the total of all elements in an array. NumPy provides the np.sum() function to add numerical values quickly.
## Example
A shop sells products worth ₹100, ₹200, and ₹300. The sum gives the total sales amount.

```python
import numpy as np

sales = np.array([100, 200, 300])

total_sales = np.sum(sales)

print("Sales:", sales)
print("Total sales:", total_sales)
```

## Mean
The mean is the average of a collection of values. It is calculated by adding all the values and dividing the total by the number of values.
### Example
A student scores 70, 80, and 90 in three tests. The mean represents the average score.

```python
import numpy as np

marks = np.array([70, 80, 90])

average = np.mean(marks)

print("Average marks:", average)
```

## Median
The median is the middle value when the data is arranged in ascending order. If the number of values is even, the median is the average of the two middle values.
### Example
Employee salaries are ₹15,000, ₹20,000, and ₹25,000. The median salary is ₹20,000.

```python
import numpy as np

salaries = np.array([15000, 20000, 25000])

median_salary = np.median(salaries)

print("Median salary:", median_salary)
```

## Variance
Variance measures how far values are spread out from their mean. A small variance means values are close to the mean, while a large variance indicates greater spread.
NumPy calculates population variance by default.
### Example
Consider marks of 40, 50, and 60. Their mean is 50. Variance measures the average squared distance of the marks from 50.

```python
import numpy as np

marks = np.array([40, 50, 60])

variance = np.var(marks)

print("Variance:", variance)
```

## Standard Deviation
Standard deviation measures how much values typically vary around their mean. It is the square root of the variance and is expressed in the same units as the original data.
### Example
If the average test mark is 50, standard deviation indicates how spread out the marks are around 50.

```python
import numpy as np

marks = np.array([40, 50, 60])

std_dev = np.std(marks)

print("Standard deviation:", std_dev)
```

## Minimum
The minimum is the smallest value in an array. NumPy provides np.min() to identify it.
### Example
A company records daily sales of ₹500, ₹300, and ₹700. The minimum identifies the lowest sales amount.

```python
import numpy as np

sales = np.array([500, 300, 700])

print("Minimum sales:", np.min(sales))
```

## Maximum
The maximum is the largest value in an array. The np.max() function returns that value.
### Example
For daily sales of ₹500, ₹300, and ₹700, the maximum identifies the highest sales amount.

```python
import numpy as np

sales = np.array([500, 300, 700])

print("Maximum sales:", np.max(sales))
```

## Argmin
np.argmin() returns the index of the smallest element in an array, rather than returning the smallest value itself.
### Example
A store has prices of ₹300, ₹150, and ₹250. Argmin identifies the position of the lowest price

```python
import numpy as np

prices = np.array([300, 150, 250])

index = np.argmin(prices)

print("Index of minimum:", index)
print("Minimum price:", prices[index])
```

## Argmax
np.argmax() returns the index of the largest element in an array.
### Example
A model produces class scores of 0.2, 0.7, and 0.1. Argmax identifies the class with the highest score.

```python
import numpy as np

scores = np.array([0.2, 0.7, 0.1])

index = np.argmax(scores)

print("Index of highest score:", index)
```

## Percentile
A percentile tells us the value below which a specified percentage of observations falls. For example, the 50th percentile corresponds to the median.
### Example
A teacher uses the 75th percentile to understand the score threshold below which approximately 75% of students fall.

```python
import numpy as np

marks = np.array([10, 20, 30, 40, 50])

p75 = np.percentile(marks, 75)

print("75th percentile:", p75)
```

## Dot Product
The dot product multiplies corresponding elements in two equally sized one-dimensional arrays and adds the results.
### Example
A shop sells two products. Their quantities are [2, 3], and their prices are [100, 200]. The dot product calculates total revenue.

```python
import numpy as np

quantities = np.array([2, 3])
prices = np.array([100, 200])

revenue = np.dot(quantities, prices)

print("Total revenue:", revenue)
```

## Matrix Multiplication
Matrix multiplication combines rows from one matrix with columns from another matrix. The number of columns in the first matrix must equal the number of rows in the second matrix.
### Example
A business records quantities of two products across two stores and multiplies the quantity matrix by the product-price matrix to calculate revenue.

```python
import numpy as np

A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

result = np.matmul(A, B)

print(result)
```

## Transpose
The transpose of a matrix swaps its rows and columns. A matrix with shape (2, 3) becomes a matrix with shape (3, 2).
### Example
A table that stores students in rows and subjects in columns can be transposed to put subjects in rows and students in columns.

```python
import numpy as np

data = np.array([[10, 20, 30],
                 [40, 50, 60]])

transposed = data.T

print("Original shape:", data.shape)
print("Transposed shape:", transposed.shape)
print(transposed)
```

## Basic Linear Algebra Operations
Linear algebra studies vectors, matrices, and the relationships between them. NumPy provides functions for operations such as matrix multiplication, solving linear equations, finding determinants, and calculating inverses.
### Example
A business might have two unknown product quantities and two equations describing sales totals. Linear algebra can help calculate those unknown quantities.

```python
import numpy as np

A = np.array([[2, 1],
              [1, 3]])

b = np.array([5, 6])

solution = np.linalg.solve(A, b)
determinant = np.linalg.det(A)
inverse = np.linalg.inv(A)

print("Solution:", solution)
print("Determinant:", round(determinant, 2))
print("Inverse:")
print(inverse)
```
