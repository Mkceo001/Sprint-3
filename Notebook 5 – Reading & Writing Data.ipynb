## Reading & Writing Data

## CSV Files
What is a CSV file?

CSV stands for Comma-Separated Values. It stores tabular data as plain text, with each row representing a record and commas separating the values.
#### Reading a CSV file

```python
import pandas as pd
df = pd.read_csv("employees.csv")

print(df)
```

#### Writing a CSV file
Use to_csv() to save a DataFrame.

```python
df.to_csv("employees_output.csv", index=False)
```

## Excel Files
What is an Excel file?

Excel files store data in worksheets inside a workbook. They usually have the .xlsx extension.
Excel is useful when data is organized into multiple sheets, such as employee details, monthly sales, and department reports.

```python
df = pd.read_excel("employees.xlsx")

print(df)
```

### Reading a specific sheet

```python
import pandas as pd

df = pd.read_excel("employees.xlsx")

print(df)
```

#### Writing an Excel file

```python
df.to_excel(
    "employees_output.xlsx",
    index=False
)
```

### SON Files
What is JSON?

JSON stands for JavaScript Object Notation. It stores data using key-value pairs and can represent objects and lists.

```python
[
    {
        "EmployeeID": 101,
        "Name": "Arun",
        "Department": "IT",
        "Salary": 30000
    },
    {
        "EmployeeID": 102,
        "Name": "Priya",
        "Department": "HR",
        "Salary": 35000
    }
]
```

#### Reading a JSON file

```python
df = pd.read_json("employees.json")

print(df)
```

#### Writing a JSON file

```python
df.to_json(
    "employees_output.json",
    orient="records",
    indent=4
)
```

## Parquet Files
What is Parquet?

Parquet is a column-oriented file format designed for efficient storage and analytical processing of structured data.

```python
df.to_parquet(
    "employees.parquet",
    index=False
)
```

```python
df = pd.read_parquet("employees.parquet")

print(df)
```

```python
df = pd.read_parquet(
    "employees.parquet",
    columns=["Name", "Salary"]
)

print(df)
```

Pickle is a Python serialization format. It can store Python objects, including Pandas DataFrames, so they can be loaded again later.

## SQL Tables
What is SQL?

SQL stands for Structured Query Language. It is used to store, retrieve, update, and manage structured data in databases.
Pandas can connect to SQL databases and read query results directly into a DataFrame. It can also write a DataFrame into a database table.

#### Create a database and table

```python
import sqlite3
import pandas as pd

connection = sqlite3.connect("company.db")

employees = pd.DataFrame({
    "EmployeeID": [101, 102, 103, 104],
    "Name": ["Arun", "Priya", "Kumar", "Divya"],
    "Department": ["IT", "HR", "Finance", "IT"],
    "Salary": [30000, 35000, 40000, 45000]
})

employees.to_sql(
    "employees",
    connection,
    if_exists="replace",
    index=False
)
```

#### Read an entire SQL table

```python
df = pd.read_sql(
    "SELECT * FROM employees",
    connection
)

print(df)
```

#### Read selected columns

```python
df = pd.read_sql(
    "SELECT Name, Salary FROM employees",
    connection
)

print(df)
```

#### Filter data using SQL

```python
df = pd.read_sql(
    "SELECT * FROM employees WHERE Salary > 35000",
    connection
)

print(df)
```

#### Append new data to a SQL table

```python
new_employee = pd.DataFrame({
    "EmployeeID": [105],
    "Name": ["Meena"],
    "Department": ["IT"],
    "Salary": [42000]
})

new_employee.to_sql(
    "employees",
    connection,
    if_exists="append",
    index=False
)
```

### Reading Multiple Files
Sometimes a dataset is split into multiple files. For example, a company might store sales data.

```python
import pandas as pd

# January sales data
pd.DataFrame({
    "Product": ["Laptop", "Mouse"],
    "Sales": [50000, 2000]
}).to_csv("sales_january.csv", index=False)

# February sales data
pd.DataFrame({
    "Product": ["Keyboard", "Monitor"],
    "Sales": [3000, 15000]
}).to_csv("sales_february.csv", index=False)

# March sales data
pd.DataFrame({
    "Product": ["Laptop", "Headphones"],
    "Sales": [55000, 4000]
}).to_csv("sales_march.csv", index=False)

print("All three CSV files created successfully!")
```

```python
df_january = pd.read_csv("sales_january.csv")
df_february = pd.read_csv("sales_february.csv")
df_march = pd.read_csv("sales_march.csv")

print("January Sales:")
print(df_january)

print("\nFebruary Sales:")
print(df_february)

print("\nMarch Sales:")
print(df_march)
```
