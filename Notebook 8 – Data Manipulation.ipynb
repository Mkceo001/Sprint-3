# Data Manipulation

## DataFrame

```python
import pandas as pd

data = {
    "Name": ["Arun", "Priya", "Kumar", "Divya", "Arun"],
    "Age": [22, 25, 30, 28, 22],
    "Salary": [20000, 35000, 40000, 30000, 20000],
    "Department": ["IT", "HR", "IT", "Sales", "IT"]
}

df = pd.DataFrame(data)

print(df)
```

## Rename Columns
What is rename()?

The rename() method changes existing column names or index labels without changing the actual data values.
### Example:

```python
df_renamed = df.rename(columns={
    "Name": "Employee_Name",
    "Salary": "Monthly_Salary"
})

print(df_renamed)
```

## Drop Rows
What is drop() for rows?

Dropping rows means removing unwanted records from a DataFrame using their index labels.
### Example:

```python
df_without_row = df.drop(index=2)

print(df_without_row)
```

## Drop Columns
Dropping columns means removing unwanted fields from a DataFrame.
### Example 
If we only need employee names, ages, and departments, we can remove the salary column.

```python
df_without_salary = df.drop(columns=["Salary","Age"])

print(df_without_salary)
```

## Insert Columns
What is insert()?

The insert() method adds a new column at a specific position in the DataFrame.
### Example
Add an Experience column at position 2.

```python
df.insert(2, "Experience", [1, 3, 5, 2, 1])

print(df)
```

## assign()
What is assign()?

The assign() method creates new columns or updates existing columns using calculated values.
It returns a new DataFrame and leaves the original unchanged unless you assign the result back.
### Example: 
Create a new column

```python
df_new = df.assign(
    Annual_Salary=df["Salary"] * 12
)

print(df_new[["Name", "Salary", "Annual_Salary"]])
```

## apply()
What is apply()?

The apply() method applies a function to the values in a Series or along an axis of a DataFrame.
It is useful when you need custom logic that cannot be expressed easily with a simple arithmetic operation.
### Example: 
Apply a function to a column.
Suppose we want to classify employees by age.

```python
def age_category(age):
    if age < 25:
        return "Young"
    else:
        return "Adult"

df["Age_Category"] = df["Age"].apply(age_category)

print(df[["Name", "Age", "Age_Category"]])
```

## map()
What is map()?

The map() method transforms individual values in a Series using a dictionary or a function.
It is especially useful for converting categories or replacing values based on a lookup table.
### Example
Map department codes

```python
department_codes = {
    "IT": "Information Technology",
    "HR": "Human Resources",
    "Sales": "Sales Department"
}

df["Department_Name"] = df["Department"].map(department_codes)

print(df[["Name", "Department", "Department_Name"]])
```

## replace()
What is replace()?

The replace() method replaces specific values in a Series or DataFrame with new values.
Unlike map(), it leaves values that are not specified in the replacement mapping unchanged.
### Example: 
Replace one value

```python
df_replaced = df.replace("IT", "Technology")

print(df_replaced[["Name", "Department"]])
```

## Duplicate Handling
What are duplicate records?

Duplicate records are rows that repeat data already present in a dataset.
Duplicates may increase record counts, distort averages, and affect machine learning model training.

Pandas provides two main methods:
- duplicated() identifies duplicate rows.
- drop_duplicates() removes duplicate rows.
  ### Example

```python
print(df.duplicated())
```
