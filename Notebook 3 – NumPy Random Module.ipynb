## NumPy Random Module

## Random Numbers
Random numbers are values generated using a random-number generator. They are useful when we need sample data, simulations, or randomized inputs.
### Example
A company simulates customer waiting times or generates sample measurements to test an analysis program.

```python
import numpy as np

rng = np.random.default_rng()

numbers = rng.random(5)

print(numbers)
```

## Random Seed
A random seed initializes a random-number generator. Using the same seed with the same generator and sequence of operations allows us to reproduce the same random results.
### Example
Two developers want to test a machine learning experiment using the same randomly generated sample data.

```python
import numpy as np

rng1 = np.random.default_rng(42)
rng2 = np.random.default_rng(42)

values1 = rng1.integers(1, 100, size=5)
values2 = rng2.integers(1, 100, size=5)

print("First:", values1)
print("Second:", values2)
print("Same results:", np.array_equal(values1, values2))
```

## Random Integers
Random integers are whole numbers selected randomly within a specified range. In NumPy, the integers() method uses an inclusive lower bound and an exclusive upper bound.
### Example
A teacher randomly generates five numbers between 1 and 10 for a classroom activity.

```python
import numpy as np

rng = np.random.default_rng(42)

numbers = rng.integers(1, 11, size=5)

print(numbers)
```

## Random Choice
Random choice selects one or more elements from a given collection. You can select with replacement, meaning an element may be selected multiple times, or without replacement, meaning selected elements are not repeated.
### Example
A company randomly selects employees for a survey.

```python
import numpy as np

rng = np.random.default_rng(42)

employees = np.array(["Arun", "Priya", "Kumar", "Divya"])

selected = rng.choice(employees, size=2, replace=False)

print("Selected employees:", selected)
```

## Shuffle
Shuffling changes the order of elements in an array randomly. NumPy's shuffle() method modifies the array in place.
### Example
Before training a machine learning model, we may shuffle the order of training examples to reduce the influence of their original ordering.

```python
import numpy as np

rng = np.random.default_rng(42)

data = np.array([1, 2, 3, 4, 5])

rng.shuffle(data)

print("Shuffled data:", data)
```

## Permutation
Permutation returns a randomly reordered sequence or a permutation of the supplied elements. Unlike shuffle(), which changes an array in place, permutation() returns a permuted result.
### Example
A teacher wants to generate a random order for presenting five students.

```python
import numpy as np

rng = np.random.default_rng(42)

students = np.array(["A", "B", "C", "D", "E"])

order = rng.permutation(students)

print("Original:", students)
print("Random order:", order)
```
