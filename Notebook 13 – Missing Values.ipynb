# Missing Values

Missing values are data points that have no available value in a dataset. They are commonly represented by:
- NaN – Not a Number
- None – Python's representation of no value
- NaT – Not a Time, used for missing date/time values
- pd.NA – Pandas' missing-value marker

```python
import pandas as pd
import numpy as np

data = {
    "Name": ["Arun", "Priya", "Kumar", "Divya", "Ravi"],
    "Age": [25, np.nan, 30, 28, np.nan],
    "Salary": [25000, 30000, np.nan, 35000, 40000],
    "Department": ["IT", "HR", None, "IT", "HR"]
}

df = pd.DataFrame(data)

print(df)
```

## isnull() 
Identify missing values
The isnull() method checks whether each value is missing.
- Returns True if a value is missing.
- Returns False if a value is available.
  ### Example:

```python
print(df.isnull())
```

## notnull() 
Identify available values.
The notnull() method does the opposite of isnull().
- Returns True if a value is present.
- Returns False if a value is missing.
### Example:

```python
print(df["Age"].notnull())
```

## dropna() 
Remove missing values.
The dropna() method removes rows or columns containing missing values.
Use it when missing records cannot be reliably recovered and removing them will not cause significant information loss.
### Example:

```python
df_clean = df.dropna()

print(df_clean)
```

```python
df_columns = df.dropna(axis=1)

print(df_columns)
```

## fillna() 
Fill missing values
The fillna() method replaces missing values with a specified value.
Instead of deleting incomplete rows, we can often retain them by filling the missing data.
### Example:

```python
df_filled = df.fillna(0)

print(df_filled)
```

## interpolate() – Estimate missing values
The interpolate() method estimates missing numeric values using surrounding data points.
It is particularly useful for time-series data, such as temperature, sales, stock prices, and sensor readings.

#### Example:

```python
weather = pd.DataFrame({
    "Day": [1, 2, 3, 4, 5],
    "Temperature": [25, np.nan, 29, np.nan, 33]
})

weather["Temperature"] = weather["Temperature"].interpolate()

print(weather)
```

## Missing Value Handling Strategies
There is no single best method for every dataset. The right strategy depends on the type of data, the amount of missingness, and why values are missing.
