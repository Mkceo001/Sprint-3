#  Combining Datasets

## Concat() – Concatenating DataFrames
What is concat()?

The Pandas concat() function combines two or more DataFrames by placing them one below another or side by side.
Think of two Excel sheets with the same columns. You want to combine their rows into one sheet. This is a typical use case for concat().
### Example
Vertical concatenation

Imagine a company has sales records from January and February.

```python
import pandas as pd

# January sales
df_january = pd.DataFrame({
    "Product": ["Laptop", "Mouse"],
    "Sales": [50000, 1000]
})

# February sales
df_february = pd.DataFrame({
    "Product": ["Keyboard", "Monitor"],
    "Sales": [2000, 15000]
})

# Combine both datasets
df_all = pd.concat(
    [df_january, df_february],
    ignore_index=True
)

print(df_all)
```

## merge() – Combining DataFrames Using Common Keys
What is merge()?

The Pandas merge() function combines two DataFrames using a common column or key.
Think of a database with two tables:
- Customers: customer ID and customer name.
- Orders: customer ID and order amount.
### Example: 
Customer and order datasets

```python
import pandas as pd

customers = pd.DataFrame({
    "CustomerID": [101, 102, 103, 104],
    "Name": ["Arun", "Priya", "Kumar", "Divya"]
})

orders = pd.DataFrame({
    "CustomerID": [101, 103, 105],
    "Amount": [5000, 3000, 7000]
})

result = pd.merge(
    customers,
    orders,
    on="CustomerID",
    how="inner"
)

print(result)
```

### Four main merge types
- Inner merge — how="inner"
Keeps only rows with matching keys in both DataFrames.
- Left merge — how="left"
Keeps every row from the left DataFrame and adds matching data from the right.
- Right merge — how="right"
Keeps every row from the right DataFrame and adds matching data from the left.
- Outer merge — how="outer"
Keeps all keys from both DataFrames and matches the available information.

## join() – Combining DataFrames Using Indexes
What is join()?

The Pandas join() method combines DataFrames primarily using their indexes.
The main difference is:
- merge() is commonly used with matching columns or keys.
- join() is commonly used with indexes.
### Example: 
Student information and marks

```python
import pandas as pd

students = pd.DataFrame({
    "Name": ["Arun", "Priya", "Kumar"]
}, index=[101, 102, 103])

marks = pd.DataFrame({
    "Marks": [85, 92, 78]
}, index=[101, 102, 103])

result = students.join(marks)

print(result)
```

## append() – Deprecation and Alternatives
What is append()?

In older versions of Pandas, DataFrame.append() was used to add rows from one DataFrame to another.
### Example
pd.concat()
Instead of append(), use pd.concat().

```python
import pandas as pd

df1 = pd.DataFrame({
    "Name": ["Arun", "Priya"],
    "Marks": [85, 90]
})

df2 = pd.DataFrame({
    "Name": ["Kumar"],
    "Marks": [78]
})

result = pd.concat(
    [df1, df2],
    ignore_index=True
)

print(result)
```

## combine_first() – Filling Missing Values from Another DataFrame
What is combine_first()?

The combine_first() method fills missing values (NaN) in one DataFrame using available values from another DataFrame.
It matches the DataFrames by their index and column labels.
#### Simple rule:
- If the first DataFrame has a value, keep it.
- If the first DataFrame has NaN, use the corresponding value from the second DataFrame.
- If both have a value, the first DataFrame's value takes priority.
  ### Example
  Combining customer contact details
  
Suppose an organization has two customer records. The first contains some missing phone numbers and email addresses, while the second contains updated information.

```python
import pandas as pd
import numpy as np

df1 = pd.DataFrame({
    "Name": ["Arun", "Priya", "Kumar"],
    "Phone": ["9876543210", np.nan, "9123456780"],
    "Email": ["arun@gmail.com", "priya@gmail.com", np.nan]
}, index=[101, 102, 103])

df2 = pd.DataFrame({
    "Name": ["Arun Updated", "Priya", "Kumar"],
    "Phone": ["9000000001", "9888888888", "9111111111"],
    "Email": ["arun_new@gmail.com", "priya_new@gmail.com", "kumar@gmail.com"]
}, index=[101, 102, 103])

result = df1.combine_first(df2)

print(result)
```
