# Introduction to Pandas

## What is Pandas?
Pandas is a Python library used for data manipulation, data analysis, and data cleaning. It helps us organize data in rows and columns, similar to an Excel spreadsheet or a SQL table.
## Why Pandas?
Pandas makes working with structured data easier than using basic Python lists and dictionaries alone.
Consider this employee dataset:
Employee	Age	Salary

Arun	25	25,000

Priya	28	35,000

Kumar	30	40,000

Divya	24	28,000    

```python
import pandas as pd

data = {
    "Employee": ["Arun", "Priya", "Kumar", "Divya"],
    "Age": [25, 28, 30, 24],
    "Salary": [25000, 35000, 40000, 28000]
}

df = pd.DataFrame(data)

print("Employee Data:")
print(df)

print("\nAverage Salary:")
print(df["Salary"].mean())

print("\nEmployees earning more than 30000:")
print(df[df["Salary"] > 30000])
```

## What is a Series?
A Series is a one-dimensional labeled data structure in Pandas. It stores a sequence of values with an associated index.
Think of a Series as a single column in a spreadsheet.
#### Example
The salaries of four employees can be stored in one Series.

```python
import pandas as pd

salaries = pd.Series([25000, 35000, 40000, 28000])

print(salaries)
```

## What is a DataFrame?
A DataFrame is a two-dimensional labeled data structure consisting of rows and columns. Each column can contain a different data type
### example:
Index	Employee	Age	Salary	Department

0	Arun	25	25,000	IT

1	Priya	28	35,000	HR

2	Kumar	30	40,000	IT

3	Divya	24	28,000	Finance


```python
import pandas as pd

data = {
    "Employee": ["Arun", "Priya", "Kumar", "Divya"],
    "Age": [25, 28, 30, 24],
    "Salary": [25000, 35000, 40000, 28000],
    "Department": ["IT", "HR", "IT", "Finance"]
}

df = pd.DataFrame(data)

print(df)
```

## Creating DataFrames
Pandas provides several ways to create a DataFrame, depending on the structure of your data


```python
import pandas as pd

data = {
    "Name": ["Arun", "Priya", "Kumar"],
    "Age": [25, 28, 30],
    "City": ["Chennai", "Madurai", "Salem"]
}

df = pd.DataFrame(data)

print(df)
```

## Data Types in Pandas
A data type tells Python what kind of value is stored in a column or Series. Correct data types are important for calculations, comparisons, data cleaning, and Machine Learning.
### Example
- int64 - Integer numbers - 100
- float64 - Decimal numbers - 25.5


```python
import pandas as pd

data = {
    "Name": ["Arun", "Priya", "Kumar"],
    "Age": [25, 28, 30],
    "Salary": [25000.5, 35000.0, 40000.75],
    "Is_Active": [True, False, True]
}

df = pd.DataFrame(data)

print(df.dtypes)
```

## Index in Pandas
An index is the label used to identify each row in a Series or DataFrame.
By default, Pandas assigns integer labels starting from 0. However, you can use custom labels such as employee IDs or product codes.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Arun", "Priya", "Kumar"],
    "Age": [25, 28, 30]
})

print(df.index)
print(df)
```

## Columns in Pandas
Columns store the different attributes of your dataset. In an employee dataset, columns might include name, age, salary, department, and joining date.


```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Arun", "Priya", "Kumar"],
    "Age": [25, 28, 30],
    "Salary": [25000, 35000, 40000]
})

print(df.columns)
```

```python
df["Annual_Salary"] = df["Salary"] * 12

print(df[["Name", "Salary", "Annual_Salary"]])
```

## Reading Data in Pandas
Data is usually stored in external files rather than typed manually into Python. Pandas provides functions to load these files into DataFrames.
Common file formats include:
- CSV — pd.read_csv()
- Excel — pd.read_excel()
- JSON — pd.read_json()
- SQL databases — pd.read_sql()

```python
df = pd.read_csv(
    "employees.csv",
    usecols=["Employee", "Salary"]
)

print(df)
```

## Writing Data to a CSV File
CSV stands for Comma-Separated Values. It stores tabular data as plain text, with each row on a new line and values separated by commas.

```python
import pandas as pd

data = {
    "Name": ["Arun", "Priya", "Kumar"],
    "Age": [25, 28, 30],
    "Salary": [25000, 35000, 40000]
}

df = pd.DataFrame(data)

# Write data to a CSV file
df.to_csv("employees.csv", index=False)

print("Data saved successfully!")
```

## Writing Data to an Excel File

```python
import pandas as pd

data = {
    "Name": ["Arun", "Priya", "Kumar"],
    "Age": [25, 28, 30],
    "Salary": [25000, 35000, 40000]
}

df = pd.DataFrame(data)

df.to_excel("employees.xlsx", index=False)

print("Excel file saved successfully!")
```

## Writing Data to a JSON File

```python
import pandas as pd

data = {
    "Name": ["Arun", "Priya", "Kumar"],
    "Age": [25, 28, 30],
    "Salary": [25000, 35000, 40000]
}

df = pd.DataFrame(data)

df.to_json(
    "employees.json",
    orient="records",
    indent=4
)

print("JSON file saved successfully!")
```
