## Calculate Statistical Measures

Given the values [10, 20, 30, 40, 50], calculate the mean, median, variance, standard deviation, minimum, maximum, and 75th percentile.
### Example:

```python
import numpy as np

data = np.array([10, 20, 30, 40, 50])

print("Mean:", np.mean(data))
print("Median:", np.median(data))
print("Variance:", np.var(data))
print("Standard Deviation:", np.std(data))
print("Minimum:", np.min(data))
print("Maximum:", np.max(data))
print("75th Percentile:", np.percentile(data, 75))
```

## Read a CSV File
Create an employee DataFrame, save it as a CSV file, and read it back using Pandas.
### Example:


```python
import pandas as pd

employees = pd.DataFrame({
    "EmployeeID": [101, 102, 103],
    "Name": ["Arun", "Bala", "Kumar"],
    "Salary": [25000, 30000, 35000]
})

employees.to_csv("employees.csv", index=False)

df = pd.read_csv("employees.csv")

print(df)
```

## Merge Two DataFrames
Create two DataFrames: one containing employee names and IDs, and another containing employee IDs and departments. Merge them using EmployeeID.
### Example:

```python
import pandas as pd

employees = pd.DataFrame({
    "EmployeeID": [101, 102, 103],
    "Name": ["Arun", "Bala", "Kumar"]
})

departments = pd.DataFrame({
    "EmployeeID": [101, 102, 103],
    "Department": ["IT", "HR", "Finance"]
})

merged_df = pd.merge(
    employees,
    departments,
    on="EmployeeID",
    how="inner"
)

print(merged_df)
```

## GroupBy and Aggregation
Calculate the total salary, average salary, and number of employees in each department.
### Example:

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Arun", "Bala", "Kumar", "Priya", "Divya"],
    "Department": ["IT", "HR", "IT", "HR", "IT"],
    "Salary": [30000, 25000, 40000, 28000, 35000]
})

result = df.groupby("Department").agg(
    Total_Salary=("Salary", "sum"),
    Average_Salary=("Salary", "mean"),
    Employee_Count=("Name", "count")
)

print(result)
```

## Filter Rows Using Multiple Conditions
From the following employee dataset, select employees who belong to the IT department and earn more than ₹30,000.
### Example:

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Arun", "Bala", "Kumar", "Priya"],
    "Department": ["IT", "HR", "IT", "IT"],
    "Salary": [35000, 25000, 45000, 28000]
})

filtered_df = df[
    (df["Department"] == "IT") &
    (df["Salary"] > 30000)
]

print(filtered_df)
```

## Handle Missing Values
 Create a DataFrame with missing salary and age values. Identify missing values, fill missing salaries with the salary mean, and remove rows where age is missing.
### Example :

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Name": ["Arun", "Bala", "Kumar", "Priya"],
    "Age": [25, np.nan, 30, 28],
    "Salary": [30000, 40000, np.nan, 35000]
})

print("Missing values:")
print(df.isnull().sum())

df["Salary"] = df["Salary"].fillna(df["Salary"].mean())

df = df.dropna(subset=["Age"])

print("\nCleaned DataFrame:")
print(df)
```

## Optimize DataFrame Memory Usage
 Create a DataFrame containing employee IDs, names, departments, and salaries. Check its memory usage, then optimize suitable data types.
### Example:

```python
import pandas as pd

df = pd.DataFrame({
    "EmployeeID": [101, 102, 103, 104],
    "Name": ["Arun", "Bala", "Kumar", "Priya"],
    "Department": ["IT", "HR", "IT", "HR"],
    "Salary": [30000, 25000, 40000, 35000]
})

print("Before optimization:")
print(df.memory_usage(deep=True))
print("Total bytes:", df.memory_usage(deep=True).sum())

df["EmployeeID"] = df["EmployeeID"].astype("int32")
df["Salary"] = df["Salary"].astype("int32")
df["Department"] = df["Department"].astype("category")

print("\nAfter optimization:")
print(df.memory_usage(deep=True))
print("Total bytes:", df.memory_usage(deep=True).sum())
```
